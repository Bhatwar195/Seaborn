## Seaborn Violin plot 

A violin plot combines the features of a **box plot** and a **KDE plot**.
It shows the data distribution, density, median, and quartiles.

### Import Required Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)
```

```python
df=sns.load_dataset('tips')
```

### Basic Violin Plot

```python
sns.violinplot(x="day", y="total_bill", data=df)
plt.title("Total Bill Distribution by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

### Violin Plot with Hue

```python
sns.violinplot(
    x="day",
    y="total_bill",
    hue="sex",
    data=df,
    split=True
)
plt.title("Total Bill Distribution by Day and Gender")
plt.show()
```

### Violin Plot with Inner Quartiles

```python
sns.violinplot(
    x="day",
    y="total_bill",
    data=df,
    inner="quartile"
)
plt.title("Violin Plot with Quartiles")
plt.show()
```

### Horizontal Violin Plot

```python
sns.violinplot(y="total_bill", data=df)

plt.title("Horizontal Violin Plot of Total Bill")
plt.show()
```
