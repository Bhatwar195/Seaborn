## Seaborn Pair plot 

A pair plot shows **pairwise relationships** between numerical variables   
in a dataset using scatter plots and histograms/KDE on the diagonal.

```python
import pandas as pd
import seaborn as sns 
import matplotlib.pyplot as plt
import matplotlib.style as style 
sns.set_theme(style="whitegrid")
```

#### Load Dataset 

```python
tips=sns.load_dataset('tips')
tips.head()
```

### Basic Pair Plot

```python
sns.pairplot(tips)
plt.show()
```

### Pair Plot with Hue

```python
sns.pairplot(tips,hue='sex')
plt.show()
```

### Pair Plot with KDE

```python
sns.pairplot(
    tips,
    hue="sex",
    diag_kind="kde"
)
plt.show()
```

### Pair Plot with Selected Columns

```python
sns.pairplot(tips,vars=['tip','total_bill'],hue='sex')
```
