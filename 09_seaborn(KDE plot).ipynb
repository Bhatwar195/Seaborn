### Seaborn KDE Plot 

A KDE plot visualizes the **probability density** of a continuous variable  
using a smooth curve instead of discrete bars.

```python
import pandas as pd
import seaborn as sns 
import matplotlib.pyplot as plt
import matplotlib.style as style
style.use('ggplot')
```

### Load Tips Dataset 

```python
tips=sns.load_dataset('tips')
tips.head()
```

### Basic KDE Plot 

```python
sns.kdeplot(tips['total_bill'],fill=True)
plt.title("KDE Plot of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Density")
plt.show()
```

### KDE Plot with hue

```python
sns.kdeplot(data=tips,x='total_bill',hue='sex',fill=True)
plt.title("KDE Plot of Total Bill by Gender")
plt.show()
```

### KDE Contour Plot 

In Seaborn KDE, a contour plot is used to visualize the joint density distribution of two numerical variables using contour  
lines instead of a filled surface. Each contour line represents areas with equal density values.

```python
sns.kdeplot(data=tips,x='total_bill',y='tip',fill=True)
plt.show()
```

### KDE Plot on Days Vs total_bill

```python
sns.kdeplot(data=tips,x='total_bill',hue='day')
plt.show()
```
